## Section 1: Manual Trace and Dry Run

1. Implement insertion and selection sort.
2. Dry-run both algorithms on a small list `[6, 3, 5, 1, 8, 2]`.
3. Output the internal steps clearly to understand the sorting process.

```python
def insertion_sort_verbose(arr):
    print(f"Initial: {arr}")
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        print(f"\nInserting {key} at position {i}")
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            print(f"  Shift {arr[j]} right: {arr}")
            j -= 1
        arr[j + 1] = key
        print(f"  Place {key} at position {j+1}: {arr}")
    print(f"Sorted: {arr}")
    return arr

def selection_sort_verbose(arr):
    print(f"Initial: {arr}")
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        print(f"\nSelect min {arr[min_idx]} from {arr[i:]} and swap with {arr[i]}")
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        print(f"  After swap: {arr}")
    print(f"Sorted: {arr}")
    return arr

# Try them here
arr1 = [6, 3, 5, 1, 8, 2]
print("Insertion Sort Steps:")
insertion_sort_verbose(arr1.copy())

print("\nSelection Sort Steps:")
selection_sort_verbose(arr1.copy())
```

## Section 2: Time Complexity Analysis with matplotlib

Generate arrays of increasing size and measure the execution time of both sorts. Plot the results.

```python
import random
import time
import matplotlib.pyplot as plt

def insertion_sort(arr):
    arr = arr.copy()
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

def selection_sort(arr):
    arr = arr.copy()
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

def time_insertion_sort(arr):
    start = time.time()
    insertion_sort(arr)
    return time.time() - start

def time_selection_sort(arr):
    start = time.time()
    selection_sort(arr)
    return time.time() - start

sizes = list(range(100, 1100, 100))
insertion_times = []
selection_times = []

for size in sizes:
    test_data = [random.randint(1, 10000) for _ in range(size)]
    insertion_times.append(time_insertion_sort(test_data))
    selection_times.append(time_selection_sort(test_data))

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(sizes, insertion_times, label="Insertion Sort", marker='o')
plt.plot(sizes, selection_times, label="Selection Sort", marker='x')
plt.xlabel("Input Size")
plt.ylabel("Time (seconds)")
plt.title("Execution Time Comparison")
plt.legend()
plt.grid(True)
plt.show()
```

## Section 3: Real-World Data Use Case

You are given a CSV of product reviews where each review has an integer rating. Your job is to sort reviews by ratings using both algorithms.

```python
import pandas as pd
random.seed(42)
# Simulated reviews
reviews = pd.DataFrame({
    'product_id': [f'P{i}' for i in range(1, 21)],
    'rating': random.choices(range(1, 6), k=20)
})
print("Original Data:")
print(reviews)

def insertion_sort_reviews(arr):
    arr = arr.copy()
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j][1] > key[1]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

def selection_sort_reviews(arr):
    arr = arr.copy()
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j][1] < arr[min_idx][1]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# Sort by rating using insertion sort
sorted_reviews_ins = reviews.copy()
sorted_reviews_ins_list = sorted_reviews_ins.values.tolist()
sorted_reviews_ins_list = insertion_sort_reviews(sorted_reviews_ins_list)
print("\nSorted by Insertion Sort:")
print(pd.DataFrame(sorted_reviews_ins_list, columns=reviews.columns))

# Sort by rating using selection sort
sorted_reviews_sel = reviews.copy()
sorted_reviews_sel_list = sorted_reviews_sel.values.tolist()
sorted_reviews_sel_list = selection_sort_reviews(sorted_reviews_sel_list)
print("\nSorted by Selection Sort:")
print(pd.DataFrame(sorted_reviews_sel_list, columns=reviews.columns))
```

## Section 4: Advanced Challenge - Adaptive Insertion Sort

Enhance insertion sort using binary search for position lookup to reduce comparisons. 
Compare the performance with your previous implementation. Hint: Use `bisect` module for binary search.

```python
import bisect

def insertion_sort_binary(arr):
    arr = arr.copy()
    for i in range(1, len(arr)):
        key = arr[i]
        # Find index to insert using binary search
        idx = bisect.bisect_left(arr, key, 0, i)
        # Move all elements right to make space
        arr = arr[:idx] + [key] + arr[idx:i] + arr[i+1:]
    return arr

# Test and compare
sample = [random.randint(1, 1000) for _ in range(1000)]
start = time.time()
insertion_sort(sample.copy())
print(f"Traditional Insertion Sort Time: {time.time() - start:.4f}s")

start = time.time()
insertion_sort_binary(sample.copy())
print(f"Binary Insertion Sort Time: {time.time() - start:.4f}s")
```

## Wrap-up Questions

1. **Which algorithm is more adaptive to sorted or nearly-sorted data?**  
   *Insertion sort* is more adaptive because it runs in nearly linear time for nearly-sorted data.

2. **Why might selection sort be useful when writing to memory is expensive?**  
   Selection sort minimizes the number of writes (swaps), making it suitable when memory writes are costly.

3. **Under what condition would insertion sort be your preferred choice in real systems?**  
   When the data is small or nearly sorted, or when stability is required.

4. **Discuss the trade-offs in terms of comparisons vs. swaps between these two.**  
   Insertion sort does fewer swaps but may do more comparisons in the worst case. Selection sort always does O(n²) comparisons but only O(n) swaps.
