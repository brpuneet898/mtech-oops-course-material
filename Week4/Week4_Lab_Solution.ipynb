## Part 1: UML Modeling from Codebase (Solution)

**Example:**
- File: `bad_srp/bad_srp.py`
- Class: `Journal` handles both journal entries and file writing (SRP Violation)

**Bad UML Class Design:**
```plaintext
----------------------
|     Journal        |
----------------------
| - entries: List[str] |
----------------------
| + add_entry()     |
| + save_to_file()  |
----------------------
```
**SRP violated by combining data management and persistence.**

## Part 2: SOLID Refactoring (Solution)

**Refactor 1 (SRP fix):** Split persistence into a new class `PersistenceManager`

```python
class Journal:
    def __init__(self):
        self.entries = []

    def add_entry(self, entry):
        self.entries.append(entry)

class PersistenceManager:
    def save_to_file(self, journal, filename):
        with open(filename, 'w') as f:
            f.write('\n'.join(journal.entries))
```

**New UML:**
```plaintext
+---------------------+       +--------------------------+
|      Journal        |       |   PersistenceManager     |
+---------------------+       +--------------------------+
| - entries: list     |       |                          |
+---------------------+       +--------------------------+
| + add_entry()       |       | + save_to_file()         |
+---------------------+       +--------------------------+
```
