SOLID Design, UML & Test-Driven Development

## Objective
Apply SOLID principles, model using UML class diagrams, and build a small component using TDD. Students will reverse-engineer design from code, improve it, and apply testing using real-world patterns.

**Duration:** 2 hours

## Part 1: UML Modeling from Codebase

### Task 1.1: Clone & Analyze the Repo
> Clone the following repo:

```bash
git clone https://github.com/beatriz-cantilho/solid-principles-python.git
```

- Navigate to `bad_srp/bad_srp.py` and other `bad_*.py` files.
- Identify major classes and relationships.
- Look for poor class design and SOLID violations.

### Task 1.2: Draw UML Class Diagram

- Use **draw.io**, **UMLet**, or pen-paper to draw a class diagram.
- Include:
  - Attributes (visibility, type)
  - Methods (visibility, return type)
  - Relationships: association, inheritance, composition

**Deliverable:** `bad_design_class_diagram.png` or `.drawio` file

## Part 2: Apply SOLID Principles

### Task 2.1: Refactor Code

- Apply **SRP, DIP, LSP** in at least 3 places.
- Extract responsibilities into smaller classes.
- Use interfaces/abstractions for dependencies.

### Task 2.2: UML (Post-refactor)

- Draw a new class diagram after refactor.
- Highlight changes (e.g., color-coded differences).

**Deliverables:**
- `refactored_code/` folder
- `uml_refactored.png`

## Part 3: TDD Cycle & Testing

### Task 3.1: Identify a Component to Test

Pick one logic component (e.g., `ReportGenerator`, `Repository`, etc.)

### Task 3.2: TDD Implementation

- Follow Red-Green-Refactor:
  - Write test first (fail)
  - Write minimum logic to pass
  - Refactor if needed
- Use `unittest` or `pytest`

**Deliverables:**

- `tests/` directory with 3+ unit tests
- Final working class with SOLID adherence

Answer the following in a markdown cell below:
- Which SOLID principle was hardest to apply and why?
- What UML relationship was tricky to model?
- How did TDD improve/refine your design?
