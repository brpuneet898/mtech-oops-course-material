## Section 2: Python Sequences & Data Model

### Task 2.1: Implement Extended FrenchDeck
Implement the classic FrenchDeck with enhancements.


````python
import collections, random

Card = collections.namedtuple('Card', ['rank', 'suit'])

class FrenchDeck:
    ranks = [str(n) for n in range(2, 11)] + list('JQKA')
    suits = 'spades diamonds clubs hearts'.split()

    def __init__(self):
        self._cards = [Card(rank, suit) for suit in self.suits
                                        for rank in self.ranks]

    def __len__(self):
        return len(self._cards)

    def __getitem__(self, position):
        return self._cards[position]

    def __setitem__(self, position, value):
        self._cards[position] = value

    def __contains__(self, card):
        return card in self._cards

    def __str__(self):
        return f"Deck of {len(self)} cards"

    def __repr__(self):
        return f"FrenchDeck({self._cards!r})"
````

### Task 2.2: Unit Test FrenchDeck

````python
deck = FrenchDeck()
assert len(deck) == 52
assert deck[0] == Card('2', 'spades')
assert Card('Q', 'hearts') in deck
print(deck[:3])  # slicing
random.shuffle(deck)
print(deck[:3])  # confirm shuffle
print("All tests passed.")
````

## Section 3: Practical Type Hints and Static Checking

### Task 3.1: Annotate FrenchDeck with Type Hints and Check with mypy


````python
# Example based on the type hinting concept. 
from typing import List, Tuple

def count_suits(deck: FrenchDeck) -> dict[str, int]:
    count = {suit: 0 for suit in FrenchDeck.suits}
    for card in deck:
        count[card.suit] += 1
    return count

print(count_suits(FrenchDeck()))
````

## Section 4: Data Model & Type Hints

### Task 4.1: Build a Type-Hinted Vector2D Class


````python
import math
from typing import Any

class Vector2D:
    def __init__(self, x: float = 0, y: float = 0):
        self.x = x
        self.y = y

    def __repr__(self) -> str:
        return f"Vector2D({self.x!r}, {self.y!r})"

    def __abs__(self) -> float:
        return math.hypot(self.x, self.y)

    def __add__(self, other: 'Vector2D') -> 'Vector2D':
        return Vector2D(self.x + other.x, self.y + other.y)

    def __mul__(self, scalar: float) -> 'Vector2D':
        return Vector2D(self.x * scalar, self.y * scalar)

    def __bool__(self) -> bool:
        return bool(self.x or self.y)

    def __eq__(self, other: Any) -> bool:
        return isinstance(other, Vector2D) and self.x == other.x and self.y == other.y
````

````python
v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)
print(v1 + v2)
print(abs(v1))
print(v1 * 2)
print(bool(Vector2D()))
print(v1 == Vector2D(3, 4))
````
