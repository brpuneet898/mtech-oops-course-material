# Complex Number System

### Concepts Covered - 

1. Custom class
2. Operator overloading
3. Polar conversion
4. Equality
5. Immutability
6. `@classmethod`
7. `__eq__` 
8. `__hash__` 
9. Complex namespaces

## A: Basic Arithmetic

- Define `Complex` class with attributes `real`, `imag`
- Implement 
    - `__init__`, `__str__`, `__repr__`
    - Arithmetic: `__add__`, `__sub__`, `__mul__`, `__truediv__`
    - `__eq__`, `__abs__`, `__neg__`, `__pow__`

## B: Class Features & Namespace Exploration

- Use `@classmethod` to create `Complex` from polar coordinates
- `from_string(cls, s: str)` → Parse strings like `"3+4j"`, `"2-5j"`
- Add `to_polar()` and `to_tuple()` representations
- Add private attributes `_real`, `_imag` with read-only properties

## C: Design Problem

- **Problem**: Redesign the class to make it immutable after initialization.
- Implement an `ImmutableComplex` subclass where once `real` and `imag` are set, they can't be changed.

## D: Extension

- Create a `ComplexVector` class that stores a list of complex numbers
    - Support vector addition, dot product, and magnitude
    - Normalize a complex vector
