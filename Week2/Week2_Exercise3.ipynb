# Bounding Box Geometry & Spatial Design

### Concepts Covered:

- Geometry reasoning
- Class Relations
- Point containment
- Intersection
- Intersection
- Abstraction
- 2D Design

## A: Base Implementation

- `BoundingBox` with `x_min`, `y_min`, `x_max`, `y_max`
- Methods:
    - `__init__`, `__str__`, `area()`, `perimeter()`
    - `contains_point(x, y)`
    - `intersects_with(other)`
    - `expand_to_include(x, y)` or another bounding box

## B: Robustness and Validation

- Validate input: Ensure `x_min <= x_max`, else raise `ValueError`
- Add `__eq__`, `__repr__`, and `__contains__`(Pythonic containment)
- Add `__bool__` to return `False` for degenerate boxes (area = 0)

## C: Hierarchical Design

- Implement a `Point` class. Modify `BoundingBox` to operate on `Point` objects instead of raw tuples. Add:
    - `Point`: With `__str__`, `distance_to(other)`, `__eq__`, `__hash__`
    - `BoundingBox` now has methods like `contains(Point)` and `intersects(BoundingBox)`

## D: BoundingBoxSet

- A class to manage multiple bounding boxes. Add:
    - Method to find union of all boxes
    - Return the smallest box enclosing all boxes
    - Method to count overlapping pairs

## E: Data Visualization

- Plot points and bounding boxes using `matplotlib`
- Shade overlapping regions differently
