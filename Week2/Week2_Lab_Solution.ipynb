## Part 1: Warm-Up 

### Task 1.1: Object Identity vs Equality

We create a `Book` class and override `__eq__` so that two books are equal if their ISBNs match. We demonstrate the difference between `is` (identity) and `==` (equality).

```python
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn

    def __eq__(self, other):
        if isinstance(other, Book):
            return self.isbn == other.isbn
        return False

book1 = Book("Python 101", "Alice", "123456")
book2 = Book("Python Basics", "Alice", "123456")
book3 = book1

print("book1 == book2:", book1 == book2)  # True
print("book1 is book2:", book1 is book2)  # False
print("book1 is book3:", book1 is book3)  # True
print("id(book1):", id(book1))
print("id(book2):", id(book2))
print("id(book3):", id(book3))
```

### Task 1.2: Namespaces and LEGB Rule

This example demonstrates the LEGB rule (Local, Enclosing, Global, Built-in) using nested functions.

```python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print("Inner:", x)
    inner()
    print("Outer:", x)

outer()
print("Global:", x)
```

## Part 2: Advanced Vector Class 

### Task 2.1: 2D Vector Class

A 2D vector class with operator overloads and a magnitude method.

```python
import math

class Vector2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector2D(x={self.x}, y={self.y})"

    def __add__(self, other):
        return Vector2D(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector2D(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector2D(self.x * scalar, self.y * scalar)

    def __truediv__(self, scalar):
        return Vector2D(self.x / scalar, self.y / scalar)

    def magnitude(self):
        return math.sqrt(self.x ** 2 + self.y ** 2)

v1 = Vector2D(3, 4)
v2 = Vector2D(1, 2)
print("v1 + v2 =", v1 + v2)
print("v1 - v2 =", v1 - v2)
print("v1 * 2 =", v1 * 2)
print("v1 / 2 =", v1 / 2)
print("Magnitude of v1 =", v1.magnitude())
```

### Task 2.2: n-Dimensional Vector

A general n-dimensional vector class with dot product and normalization.

```python
class VectorND:
    def __init__(self, *coords):
        self.coords = list(coords)

    def __str__(self):
        return str(tuple(self.coords))

    def __repr__(self):
        return f"VectorND{tuple(self.coords)}"

    def __add__(self, other):
        return VectorND(*[a + b for a, b in zip(self.coords, other.coords)])

    def __sub__(self, other):
        return VectorND(*[a - b for a, b in zip(self.coords, other.coords)])

    def dot(self, other):
        return sum(a * b for a, b in zip(self.coords, other.coords))

    def magnitude(self):
        return math.sqrt(sum(a ** 2 for a in self.coords))

    def normalize(self):
        mag = self.magnitude()
        return VectorND(*[a / mag for a in self.coords])

v3 = VectorND(1, 2, 3)
v4 = VectorND(4, 5, 6)
print("v3 + v4 =", v3 + v4)
print("Dot product:", v3.dot(v4))
print("Normalized v3:", v3.normalize())
```

## Part 3: 2D Geometry Engine 

### Task 3.1: Point2D Class

A simple 2D point class with a distance method.

```python
class Point2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def distance_to(self, other):
        return math.hypot(self.x - other.x, self.y - other.y)

p1 = Point2D(0, 0)
p2 = Point2D(3, 4)
print(p1)
print("Distance:", p1.distance_to(p2))
```

### Task 3.2: Line2D Class

A line defined by two points, with methods for length and orientation.

```python
class Line2D:
    def __init__(self, p1, p2):
        self.p1 = p1
        self.p2 = p2

    def length(self):
        return self.p1.distance_to(self.p2)

    def is_horizontal(self):
        return self.p1.y == self.p2.y

    def is_vertical(self):
        return self.p1.x == self.p2.x

line = Line2D(Point2D(0, 0), Point2D(0, 5))
print("Length:", line.length())
print("Is vertical?", line.is_vertical())
print("Is horizontal?", line.is_horizontal())
```

### Task 3.3: Polygon Class

A polygon defined by a list of points. Computes perimeter and checks if closed.

```python
class Polygon:
    def __init__(self, *points):
        self.points = list(points)

    def is_closed(self):
        return self.points[0].x == self.points[-1].x and self.points[0].y == self.points[-1].y

    def perimeter(self):
        peri = 0
        for i in range(len(self.points) - 1):
            peri += self.points[i].distance_to(self.points[i+1])
        if not self.is_closed():
            peri += self.points[-1].distance_to(self.points[0])
        return peri

triangle = Polygon(Point2D(0, 0), Point2D(3, 0), Point2D(3, 4), Point2D(0, 0))
print("Perimeter:", triangle.perimeter())
print("Closed?", triangle.is_closed())
```

## Additional: Singleton Pattern

A singleton logger class using `__new__` to ensure only one instance exists.

```python
class SingletonLogger:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(SingletonLogger, cls).__new__(cls)
            print("New instance created")
        return cls._instance

logger1 = SingletonLogger()
logger2 = SingletonLogger()
print("Same instance?", logger1 is logger2)
```
