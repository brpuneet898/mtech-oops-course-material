## Dynamic Programming

### 1. Fibonacci Number (Top-Down with Memoization)

Given an integer n, compute the n-th Fibonacci number using Dynamic Programming.

```python
def fibonacci(n, memo={}):
    if n <= 1:
        return n
    if n in memo:
        return memo[n]
    memo[n] = fibonacci(n-1, memo) + fibonacci(n-2, memo)
    return memo[n]

print(fibonacci(6))
```

### 2. Climbing Stairs (Bottom-Up DP)

You can climb 1 or 2 steps at a time. Find number of distinct ways to reach the top.

```python
def climbStairs(n):
    if n <= 2:
        return n
    dp = [0]*(n+1)
    dp[1], dp[2] = 1, 2
    for i in range(3, n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

print(climbStairs(5))
```

### 3. Minimum Cost to Reach End (1D DP)

Given cost array where cost[i] is the cost to step on that stair, find minimum cost to reach top.

```python
def minCostClimbingStairs(cost):
    n = len(cost)
    dp = [0]*(n+1)
    for i in range(2, n+1):
        dp[i] = min(dp[i-1] + cost[i-1], dp[i-2] + cost[i-2])
    return dp[n]


print(minCostClimbingStairs([1, 100, 1, 1, 1, 100, 1]))
```

### 4. Coin Change (Count Ways, Unbounded DP)

Given coins of different denominations and a total amount, count the number of ways to make that amount.

```python
def coin_change_ways(coins, amount):
    dp = [0] * (amount+1)
    dp[0] = 1

    for coin in coins:
        for x in range(coin, amount+1):
            dp[x] += dp[x-coin]

    return dp[amount]

print(coin_change_ways([1,2,5], 5))
```

### 5. Longest Common Subsequence (LCS)

Given two strings, find the length of their longest common subsequence.

```python
def lcs(text1, text2):
    n, m = len(text1), len(text2)
    dp = [[0]*(m+1) for _ in range(n+1)]

    for i in range(1, n+1):
        for j in range(1, m+1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = 1 + dp[i-1][j-1]
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])

    return dp[n][m]

print(lcs("abcde", "ace"))
```

### 6. Edit Distance (Minimum Operations)

Convert one string to another using insert, delete, or replace operations.

```python
def edit_distance(s1, s2):
    n, m = len(s1), len(s2)
    dp = [[0]*(m+1) for _ in range(n+1)]

    for i in range(n+1):
        dp[i][0] = i
    for j in range(m+1):
        dp[0][j] = j

    for i in range(1, n+1):
        for j in range(1, m+1):
            if s1[i-1] == s2[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(dp[i-1][j],
                                   dp[i][j-1],
                                   dp[i-1][j-1])
    return dp[n][m]


print(edit_distance("horse", "ros"))
```

### 7. Matrix Chain Multiplication

Find minimum cost of multiplying chain of matrices with given dimensions.

```python
def matrix_chain_multiplication(arr):
    n = len(arr)
    dp = [[0]*n for _ in range(n)]

    for L in range(2, n):
        for i in range(1, n-L+1):
            j = i + L - 1
            dp[i][j] = float('inf')
            for k in range(i, j):
                cost = dp[i][k] + dp[k+1][j] + arr[i-1]*arr[k]*arr[j]
                dp[i][j] = min(dp[i][j], cost)
    return dp[1][n-1]

print(matrix_chain_multiplication([1, 2, 3, 4]))
```

### 8. Boolean Parenthesization Problem

Count the number of ways to parenthesize a boolean expression to evaluate to True.

```python
def count_true_ways(expr, ops):
    n = len(expr)
    T = [[0]*n for _ in range(n)]
    F = [[0]*n for _ in range(n)]

    for i in range(n):
        if expr[i] == 'T':
            T[i][i] = 1
        else:
            F[i][i] = 1

    for gap in range(1, n):
        for i in range(n-gap):
            j = i + gap
            for k in range(i, j):
                op = ops[k]
                total_ik = T[i][k] + F[i][k]
                total_kj = T[k+1][j] + F[k+1][j]

                if op == '&':
                    T[i][j] += T[i][k] * T[k+1][j]
                    F[i][j] += total_ik * total_kj - T[i][k] * T[k+1][j]
                elif op == '|':
                    F[i][j] += F[i][k] * F[k+1][j]
                    T[i][j] += total_ik * total_kj - F[i][k] * F[k+1][j]
                else:  # '^'
                    T[i][j] += T[i][k] * F[k+1][j] + F[i][k] * T[k+1][j]
                    F[i][j] += T[i][k] * T[k+1][j] + F[i][k] * F[k+1][j]

    return T[0][n-1]

print(count_true_ways("TFT", "^&"))
```

### 9. Palindromic Partitioning (Min Cuts)

Partition a string into minimum number of palindromic substrings.

```python
def min_palindrome_cuts(s):
    n = len(s)
    dp = [0]*n
    pal = [[False]*n for _ in range(n)]

    for i in range(n):
        min_cut = i
        for j in range(i+1):
            if s[j] == s[i] and (i-j <= 1 or pal[j+1][i-1]):
                pal[j][i] = True
                min_cut = 0 if j == 0 else min(min_cut, dp[j-1] + 1)
        dp[i] = min_cut
    return dp[-1]

print(min_palindrome_cuts("aab"))
```
