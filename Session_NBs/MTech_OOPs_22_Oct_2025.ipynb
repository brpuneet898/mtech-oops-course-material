## LIS (Longest Increasing Subsequence)

### 1. Find the Length of LIS

Given an array of integers, return the length of the longest strictly increasing subsequence.

```python
def lengthOfLIS(nums):
    n = len(nums)
    dp = [1] * n

    for i in range(n):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)

print(lengthOfLIS([10, 9, 2, 5, 3, 7, 101, 18]))
```

### 2. Actual LIS Sequence (Reconstruct the subsequence)

Return not just the length but the actual longest increasing subsequence.

```python
def lis_sequence(nums):
    n = len(nums)
    dp = [1] * n
    parent = [-1] * n

    max_len = 0
    max_index = 0

    for i in range(n):
        for j in range(i):
            if nums[i] > nums[j] and dp[i] < dp[j] + 1:
                dp[i] = dp[j] + 1
                parent[i] = j
        if dp[i] > max_len:
            max_len = dp[i]
            max_index = i

    seq = []
    while max_index != -1:
        seq.append(nums[max_index])
        max_index = parent[max_index]

    return seq[::-1]

print(lis_sequence([3, 10, 2, 1, 20]))
```

### 3. Longest Bitonic Subsequence

A bitonic subsequence first increases and then decreases. Find its maximum length.

```python
def longestBitonicSubsequence(nums):
    n = len(nums)
    lis = [1] * n
    lds = [1] * n

    for i in range(n):
        for j in range(i):
            if nums[i] > nums[j]:
                lis[i] = max(lis[i], lis[j] + 1)

    for i in range(n-1, -1, -1):
        for j in range(n-1, i, -1):
            if nums[i] > nums[j]:
                lds[i] = max(lds[i], lds[j] + 1)

    max_len = 0
    for i in range(n):
        max_len = max(max_len, lis[i] + lds[i] - 1)

    return max_len

print(longestBitonicSubsequence([1, 11, 2, 10, 4, 5, 2, 1]))
```

## Knapsack

### 4. 0/1 Knapsack (Basic DP Table)

Given weights, values, and capacity, find the maximum value that can be achieved.

```python
def knapsack(weights, values, W):
    n = len(values)
    dp = [[0] * (W+1) for _ in range(n+1)]

    for i in range(1, n+1):
        for w in range(1, W+1):
            if weights[i-1] <= w:
                dp[i][w] = max(values[i-1] + dp[i-1][w-weights[i-1]], dp[i-1][w])
            else:
                dp[i][w] = dp[i-1][w]
    return dp[n][W]

print(knapsack([1,3,4,5], [1,4,5,7], 7))
```

### 5. Unbounded Knapsack (Infinite supply of each item)

You can include the same item multiple times.

```python
def unbounded_knapsack(weights, values, W):
    dp = [0] * (W+1)

    for w in range(W+1):
        for i in range(len(values)):
            if weights[i] <= w:
                dp[w] = max(dp[w], values[i] + dp[w-weights[i]])

    return dp[W]

print(unbounded_knapsack([2,4,6], [5,11,13], 10))
```

### 6. Count Number of Ways to Fill Knapsack Exactly

Given items, count how many ways you can achieve exactly weight W.

```python
def count_knapsack_ways(weights, W):
    dp = [0] * (W+1)
    dp[0] = 1

    for wt in weights:
        for w in range(W, wt-1, -1):
            dp[w] += dp[w-wt]

    return dp[W]

print(count_knapsack_ways([1,2,3], 4))
# Ways: [1+1+1+1], [1+1+2], [1+3], [2+2]
```
