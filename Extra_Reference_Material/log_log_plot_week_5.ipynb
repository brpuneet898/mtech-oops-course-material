```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
n_values = np.logspace(1, 6, num=100) 
comparisons = np.log2(n_values) 
```

```python
plt.figure(figsize=(8, 6))
plt.loglog(n_values, comparisons, label='Binary Search: log₂(n)', color='blue', linewidth=2)

plt.xlabel('Input Size n (log scale)')
plt.ylabel('Number of Comparisons (log scale)')
plt.title('Log-Log Plot: Binary Search Time Complexity')
plt.grid(True, which="both", ls="--", lw=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
