## Part 1: Strategy Pattern Skeleton for Sorting

### Task 1.1: Define a SortStrategy Interface

Create an abstract class or interface that all sorting strategies must implement.


```python
from abc import ABC, abstractmethod

class SortStrategy(ABC):
    @abstractmethod
    def sort(self, data: list) -> list:
        pass
```

### Task 1.2: Define a Sorter Class Using Strategy Pattern

Implement a class `Sorter` that accepts a strategy object and delegates the sorting task.


```python
class Sorter:
    def __init__(self, strategy: SortStrategy):
        self._strategy = strategy

    def set_strategy(self, strategy: SortStrategy):
        self._strategy = strategy

    def sort(self, data: list) -> list:
        return self._strategy.sort(data)
```

## Part 2: Implement Sorting Algorithms as Strategy Classes

### Task 2.1: Implement MergeSortStrategy

Write a class `MergeSortStrategy` that implements the `SortStrategy` interface.


```python
class MergeSortStrategy(SortStrategy):
    def sort(self, data: list) -> list:
        if len(data) <= 1:
            return data
        mid = len(data) // 2
        left = self.sort(data[:mid])
        right = self.sort(data[mid:])
        return self._merge(left, right)

    def _merge(self, left, right):
        result = []
        i = j = 0
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        result.extend(left[i:])
        result.extend(right[j:])
        return result
```

### Task 2.2: Implement QuickSortStrategy

Write a class `QuickSortStrategy` that implements the `SortStrategy` interface.

```python
class QuickSortStrategy(SortStrategy):
    def sort(self, data: list) -> list:
        if len(data) <= 1:
            return data
        pivot = data[0]
        lesser = [x for x in data[1:] if x <= pivot]
        greater = [x for x in data[1:] if x > pivot]
        return self.sort(lesser) + [pivot] + self.sort(greater)
```

### Task 2.3: Test Your Sorter with Different Strategies

Use the `Sorter` class to sort the same input list using both strategies and verify correctness.

```python
input_data = [5, 2, 9, 1, 5, 6]
sorter = Sorter(MergeSortStrategy())
print("MergeSort:", sorter.sort(input_data.copy()))

sorter.set_strategy(QuickSortStrategy())
print("QuickSort:", sorter.sort(input_data.copy()))
```

## Part 3: Advanced – Timsort-Inspired Sorting Strategy

### Task 3.1: Understand and Simulate Timsort Logic

Implement a **Timsort-inspired** strategy that:
- Uses insertion sort for small subarrays (size < 32)
- Falls back to merge sort on larger sorted runs

Break the list into **runs**, sort them using insertion sort, then merge them.

```python
class TimSortStrategy(SortStrategy):
    RUN = 32

    def sort(self, data: list) -> list:
        n = len(data)
        for start in range(0, n, self.RUN):
            end = min(start + self.RUN, n)
            self._insertion_sort(data, start, end)

        size = self.RUN
        while size < n:
            for left in range(0, n, 2 * size):
                mid = min(left + size, n)
                right = min((left + 2 * size), n)
                merged = self._merge(data[left:mid], data[mid:right])
                data[left:left + len(merged)] = merged
            size *= 2

        return data

    def _insertion_sort(self, data, start, end):
        for i in range(start + 1, end):
            key = data[i]
            j = i - 1
            while j >= start and data[j] > key:
                data[j + 1] = data[j]
                j -= 1
            data[j + 1] = key

    def _merge(self, left, right):
        result = []
        i = j = 0
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        result.extend(left[i:])
        result.extend(right[j:])
        return result
```

```python
class QuickSortStrategy(SortStrategy):
    def sort(self, data: list, key=lambda x: x) -> list:
        if len(data) <= 1:
            return data
        pivot = data[0]
        lesser = [x for x in data[1:] if key(x) <= key(pivot)]
        greater = [x for x in data[1:] if key(x) > key(pivot)]
        return self.sort(lesser, key) + [pivot] + self.sort(greater, key)
```
